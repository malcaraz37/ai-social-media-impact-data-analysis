### **Figura 2. Distribución de variables numéricas seleccionadas (Histogramas)**
Usando Matplotlib con plt.subplots(2, 3), representamos los histogramas con 30 divisiones para las seis variables numéricas principales, calculando su media y desviación estándar en el título (igual que en el ejercicio 8 de *Tipos de Clima*).


```python
# Figura 2: Histogramas de las variables numéricas clave
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

variables_hist = [
    ('Daily_Social_Media_Hours', 'Horas en redes sociales'),
    ('Daily_AI_Tool_Usage_Hours', 'Horas con herramientas de IA'),
    ('Sleep_Hours', 'Horas de sueño'),
    ('Mental_Health_Score', 'Puntaje de salud mental'),
    ('Academic_Performance_Score', 'Rendimiento académico'),
    ('Social_Isolation_Score', 'Aislamiento social')
]

for ax, (col, nombre) in zip(axes.flatten(), variables_hist):
    media = student_df[col].mean()
    desv = student_df[col].std()
    
    ax.hist(student_df[col], bins=30, color='gray', edgecolor='black', alpha=0.7)
    ax.set_title(f'{nombre}\nMedia: {media:.2f} | Desv. Est.: {desv:.2f}', fontsize=11)
    ax.set_xlabel('Valor')
    ax.set_ylabel('Frecuencia')

plt.suptitle('Figura 2. Distribución de variables numéricas seleccionadas', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

### **Figura 3. Matriz de correlaciones entre variables numéricas**
Calculamos la correlación de Pearson con student_df.corr(numeric_only=True) y la graficamos con sns.heatmap anotando los valores (igual que en el ejercicio 4 de *Tipos de Clima*).


```python
# Figura 3: Matriz de correlación de Pearson con Seaborn
plt.figure(figsize=(10, 8))
correlations = student_df.corr(numeric_only=True, method='pearson')

sns.heatmap(correlations, annot=True, fmt='.2f', cmap='Greys', vmin=-1, vmax=1)
plt.title('Figura 3. Matriz de correlaciones entre variables numéricas', fontsize=13, pad=12)
plt.xticks(rotation=45, ha='right')
plt.show()

# Correlaciones ordenadas con respecto al riesgo de reprobación
print('Correlación de cada variable con el riesgo de reprobación:')
display(correlations['Academic_Failure_Risk'].sort_values().to_frame(name='Correlación'))
```

### **Figura 4. Comparación de variables clave entre estudiantes sin riesgo y en riesgo (Boxplots)**
Representamos diagramas de caja (*boxplots*) con Seaborn para contrastar a los estudiantes sin riesgo (0) y en riesgo (1) en las cuatro variables clave (igual que en el ejercicio 13 de *Tipos de Clima*).


```python
# Figura 4: Boxplots comparativos entre grupos de riesgo
fig, axes = plt.subplots(1, 4, figsize=(16, 4.5))

variables_box = [
    ('Daily_Social_Media_Hours', 'Horas en redes sociales'),
    ('Sleep_Hours', 'Horas de sueño'),
    ('Mental_Health_Score', 'Salud mental'),
    ('Academic_Performance_Score', 'Rendimiento académico')
]

for ax, (col, nombre) in zip(axes, variables_box):
    sns.boxplot(x='Academic_Failure_Risk', y=col, data=student_df, ax=ax, color='lightgray')
    ax.set_title(nombre, fontsize=11)
    ax.set_xlabel('Riesgo de reprobación')
    ax.set_ylabel(nombre)

plt.suptitle('Figura 4. Comparación de variables clave entre estudiantes sin riesgo y en riesgo', fontsize=13, y=1.03)
plt.tight_layout()
plt.show()
```

### **Figura 5. Proporción de estudiantes en riesgo según su nivel de agotamiento (Gráfico de Pila / Stacked)**
Usando pd.crosstab con porcentajes normalizados por fila y .plot(kind='bar', stacked=True) (igual que en el ejercicio 14 de *Tipos de Clima*).


```python
# Figura 5: Barras apiladas (Stacked Bar) de riesgo por nivel de agotamiento
# 1. Creamos la tabla cruzada en porcentaje por fila
burnout_risk_stacked = pd.crosstab(
    student_df['Burnout_Level'],
    student_df['Academic_Failure_Risk'],
    normalize='index'
) * 100

# 2. Ordenamos las categorías de agotamiento
burnout_risk_stacked = burnout_risk_stacked.reindex(['Low', 'Moderate', 'High', 'Severe'])

# 3. Graficamos barras apiladas
burnout_risk_stacked.plot(kind='bar', stacked=True, figsize=(8, 5), color=['#d3d3d3', '#333333'], edgecolor='black')
plt.title('Figura 5. Proporción de estudiantes en riesgo según su nivel de agotamiento', fontsize=12, pad=10)
plt.xlabel('Nivel de agotamiento (Burnout Level)')
plt.ylabel('% de estudiantes')
plt.ylim(0, 105)
plt.xticks(rotation=0)
plt.legend(['Sin riesgo (0)', 'En riesgo (1)'], title='Condición')
plt.tight_layout()
plt.show()

# Mostramos la tabla numérica
display(burnout_risk_stacked.round(2))
```

## **5. Tablas estadísticas y detección de valores atípicos**
Generamos los cuadros numéricos presentados en el avance del proyecto (Tablas 2 y 3, e IQR).


### **Tabla 2. Estadísticas descriptivas de las variables numéricas**

```python
# Tabla 2: Resumen con Media, Desviación estándar, Mínimo, Mediana y Máximo
columnas_num = [
    'Age', 'Daily_Social_Media_Hours', 'Daily_AI_Tool_Usage_Hours', 'Sleep_Hours',
    'Physical_Activity_Hours', 'Mental_Health_Score', 'Physical_Health_Score',
    'Social_Isolation_Score', 'Academic_Performance_Score'
]

tabla_2 = pd.DataFrame({
    'Media': student_df[columnas_num].mean(),
    'Desv. estándar': student_df[columnas_num].std(),
    'Mínimo': student_df[columnas_num].min(),
    'Mediana': student_df[columnas_num].median(),
    'Máximo': student_df[columnas_num].max()
})

display(tabla_2.round(2))
```

### **Tabla 3. Promedios según el riesgo de reprobación**

```python
# Tabla 3: Comparación de medias agrupando por Academic_Failure_Risk
columnas_t3 = [
    'Daily_Social_Media_Hours', 'Daily_AI_Tool_Usage_Hours', 'Sleep_Hours',
    'Physical_Activity_Hours', 'Mental_Health_Score', 'Physical_Health_Score',
    'Social_Isolation_Score', 'Academic_Performance_Score'
]

tabla_3 = student_df.groupby('Academic_Failure_Risk')[columnas_t3].mean().T
tabla_3.columns = ['Sin riesgo (14,100)', 'En riesgo (900)']
display(tabla_3.round(2))
```

### **Detección de valores atípicos (Regla del Rango Intercuartílico IQR)**
Calculamos  = Q_3 - Q_1$ y contamos los registros fuera del intervalo $[Q_1 - 1.5 \cdot IQR, Q_3 + 1.5 \cdot IQR]$ (código del Anexo A del reporte).


```python
# Valores atípicos por variable (código del Anexo A del avance del proyecto)
num = student_df.select_dtypes('number')

for c in num.columns.drop('Academic_Failure_Risk'):
    q1, q3 = num[c].quantile([0.25, 0.75])
    iqr = q3 - q1
    n = ((num[c] < q1 - 1.5 * iqr) | (num[c] > q3 + 1.5 * iqr)).sum()
    print(f'{c}: {n} valores atípicos')
```

## **7. Conclusiones, Hallazgos y Recomendaciones para el Alumnado e Institución (Sección 4.5)**

### **A. Principales Hallazgos del Análisis Exploratorio (EDA)**
1. **Calidad y Consistencia de Datos:** El dataset contiene 15,000 registros completos, sin valores nulos ni filas duplicadas, lo que garantiza una base sólida para el modelado.
2. **Patrón Claro del Estudiante en Riesgo:**
   * Pasa casi el doble de tiempo en redes sociales (**7.67 hrs** vs **4.34 hrs** diarias).
   * Duerme significativamente menos (**5.35 hrs** vs **6.62 hrs** diarias).
   * Presenta menor salud mental (**59.49** vs **73.34** puntos) y mayor aislamiento social (**5.37** vs **4.24**).
   * Su rendimiento académico cae notablemente (**64.79** vs **79.18** puntos).
3. **Hallazgos Extra (Actividad Física e Intensidad de Uso de IA):**
   * **Déficit de Actividad Física:** Los estudiantes en riesgo realizan la mitad de actividad física diaria (**0.68 hrs** vs **1.28 hrs**), mostrando que el sedentarismo agrava el estrés y el bajo rendimiento.
   * **Uso Reactivo de la Inteligencia Artificial:** Quienes están en riesgo usan más horas la IA (**3.51 hrs** vs **2.53 hrs** diarias). Esto indica que muchos estudiantes recurren a la IA de forma reactiva y a última hora para intentar compensar el tiempo perdido en redes sociales o la falta de preparación, sin que esto mejore sus calificaciones.
   * **Efecto Crítico del Agotamiento:** En agotamiento *Severo* el 100% de los estudiantes está en riesgo de reprobación, y en agotamiento *Alto* el 33.3%, mientras que en niveles *Bajo* y *Moderado* el riesgo es 0%.

---

### **B. Recomendaciones Prácticas para el Alumnado (Plan de Acción para Estudiantes)**

1. **Gestión Consciente del Tiempo de Pantalla (Higiene Digital):**
   * Establecer un límite diario en aplicaciones de redes sociales (idealmente menos de 4 horas diarias) y activar el modo 'No molestar' durante los bloques de estudio.
   * Evitar el *doomscrolling* nocturno y suspender el uso del celular al menos 30 minutos antes de dormir para proteger las horas de sueño (meta: 7 horas mínimas).

2. **Uso Productivo y Ético de la Inteligencia Artificial:**
   * Utilizar la IA como un tutor personal (para resolver dudas, pedir ejemplos o autoevaluarse con preguntas de repaso) y **no como un sustituto pasivo** para resolver tareas a última hora sin comprender el contenido.

3. **Autocuidado, Movimiento y Convivencia Presencial:**
   * Realizar al menos 30 a 45 minutos de actividad física al día (caminar, deporte, gimnasio); el ejercicio ayuda a reducir los puntajes de aislamiento social y recarga la salud mental.
   * Mantener redes de apoyo con compañeros de clase y amigos para evitar el aislamiento durante semanas de alta carga académica.

4. **Identificación Temprana del Agotamiento (*Burnout*):**
   * Aprender a reconocer los síntomas del estrés crónico y agotamiento moderado/alto a tiempo, acudiendo a las tutorías y servicios de orientación académica y psicológica de la facultad antes de que impacte en las calificaciones.
